DATA CLEANING

```python
# Make a copy
cleaned_df = df.copy()
```

```python
# 29. Standardize City
cleaned_df["City"] = (
    cleaned_df["City"]
    .astype(str)
    .str.strip()
    .str.title()
)
```

```python
# 30. Standardize Destination
cleaned_df["Destination"] = (
    cleaned_df["Destination"]
    .astype(str)
    .str.strip()
    .str.title()
)
```

HANDLE MISSING VALUES

```python
# 31. Handle missing values
cleaned_df = cleaned_df.replace(
    ["NA", "N/A", "na", "n/a", "Unknown", "unknown", ""],
    pd.NA
)


for column in cleaned_df.select_dtypes(include="number").columns:
    cleaned_df[column] = cleaned_df[column].fillna(
        cleaned_df[column].median()
    )


for column in cleaned_df.select_dtypes(include="object").columns:
    if cleaned_df[column].isnull().sum() > 0:
        cleaned_df[column] = cleaned_df[column].fillna(
            cleaned_df[column].mode()[0]
        )

```

```python
# 32. Remove duplicates
cleaned_df = cleaned_df.drop_duplicates()
```

```python
# 33. Correct invalid Age values
age_median = cleaned_df.loc[
    cleaned_df["Age"].between(18, 100),
    "Age"
].median()

cleaned_df.loc[
    ~cleaned_df["Age"].between(18, 100),
    "Age"
] = age_median
```

```python
# 34. Correct invalid Customer Rating
customer_rating_median = cleaned_df.loc[
    cleaned_df["Customer_Rating"].between(0, 5),
    "Customer_Rating"
].median()

cleaned_df.loc[
    ~cleaned_df["Customer_Rating"].between(0, 5),
    "Customer_Rating"
] = customer_rating_median
```

```python
# 35. Correct invalid Hotel Rating
hotel_rating_median = cleaned_df.loc[
    cleaned_df["Hotel_Rating"].between(0, 5),
    "Hotel_Rating"
].median()

cleaned_df.loc[
    ~cleaned_df["Hotel_Rating"].between(0, 5),
    "Hotel_Rating"
] = hotel_rating_median
```

```python
# 36. Correct invalid Booking Amount
booking_median = cleaned_df.loc[
    cleaned_df["Booking_Amount"] >= 0,
    "Booking_Amount"
].median()

cleaned_df.loc[
    cleaned_df["Booking_Amount"] < 0,
    "Booking_Amount"
] = booking_median
```

```python
# 37. Correct invalid Trip Duration
trip_duration_median = cleaned_df.loc[
    cleaned_df["Trip_Duration"] > 0,
    "Trip_Duration"
].median()

cleaned_df.loc[
    cleaned_df["Trip_Duration"] <= 0,
    "Trip_Duration"
] = trip_duration_median
```

```python
# 38. Convert numeric columns to numeric
possible_numeric_columns = [
    "Age",
    "Customer_Rating",
    "Hotel_Rating",
    "Booking_Amount",
    "Trip_Duration",
    "Likes",
    "Reviews",
    "Booking_Count",
    "Cancellation_Count"
]

for column in possible_numeric_columns:
    if column in cleaned_df.columns:
        cleaned_df[column] = pd.to_numeric(
            cleaned_df[column],
            errors="coerce"
        )
```

```python
# 39. Handle NA / N/A / unknown / blank again
cleaned_df = cleaned_df.replace(
    ["NA", "N/A", "na", "n/a", "Unknown", "unknown", ""],
    pd.NA
)

for column in cleaned_df.select_dtypes(include="number").columns:
    cleaned_df[column] = cleaned_df[column].fillna(
        cleaned_df[column].median()
    )

for column in cleaned_df.select_dtypes(include="object").columns:
    if cleaned_df[column].isnull().sum() > 0:
        cleaned_df[column] = cleaned_df[column].fillna(
            cleaned_df[column].mode()[0]
        )
```
